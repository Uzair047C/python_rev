```python
class Animal:
    def speak(self):
        print("Animal makes a sound")
    def __init__(self, name):
        self.name = name
class Dog(Animal):
    def __init__(self, name):
        super().__init__(name)
    def speak(self):
        super().speak()        # Call parent's method
        print(f"{self.name} barks")
```

```python
d=Dog("Buddy")
d.speak()
```
