```python
from abc import ABC, abstractmethod
```


# abstract class


```python
class Shape(ABC):
    @abstractmethod
    def area(self):
        pass
```

# classes inherited

```python

class Rectangle(Shape):
    def __init__(self, length, width):
        self.length=length
        self.width=width
    def area(self):
        return self.length*self.width
class Circle(Shape):
    def __init__(self, radius):
        self.radius=radius
    def area(self):
        return 3.14*self.radius**2
```

# objects

```python
rec_area=Rectangle(5,4)
circle_area=Circle(3)
```

```python
print("Rectangle Area:", rec_area.area())
print("Circle Area:", circle_area.area())
```
