```python
from abc import ABC,abstractmethod
```

```python
class Animal(ABC):
    @abstractmethod
    def sound(self):
        pass
    def sleep(self):
        return "Zzz"
class Dog(Animal):
    def sound(self):
        return "Woof"
class Cat(Animal):
    def sound(self):
        return "Meow"
```

```python
cat=Cat()
dog=Dog()
```

```python
print(f"Cat says: {cat.sound()}")
print(f"Cat is sleeping: {cat.sleep()}")
```

```python
print(f"Dog says: {dog.sound()}")
print(f"Dog is sleeping: {dog.sleep()}")
```
