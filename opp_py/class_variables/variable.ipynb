```python
class Students:
    class_year = 2026
    total_students = 0

    def __init__(self, name, age):
        self.name = name
        self.age = age
        Students.total_students += 1

    def info(self):
        print(f"Name: {self.name}")
        print(f"Age: {self.age}")
        print(f"Class Year: {Students.class_year}")
 

st1 = Students("John", 20)
st2 = Students("Alice", 21)

print(f"Class Year: {Students.class_year}")
print(f"Total Students: {Students.total_students}")

st1.info()
st2.info()
```
